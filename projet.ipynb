```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("CarPriceDataset_Final.csv")
```

```python
df.head(10)
```

```python
df = df.drop_duplicates()
```

```python
df.columns.tolist()
```

```python
df = df.rename(columns={
    'Mileage ': 'Mileage',
    'Horsepower (kw)': 'Horsepower_kw',
    'Kms_driven': 'Km_driven'
})
```

```python
colonne_fuel = {
    'Gas': 'Petrol',
    'Gasoline': 'Petrol',
    'Eletric': 'Electric',
    'Hybride': 'Hybrid'
}
df['Fuel'] = df['Fuel'].replace(colonne_fuel)
```

```python
df['Mileage'] = pd.to_numeric(df['Mileage'], errors='coerce')
median_mileage = df['Mileage'].median()
df['Mileage'] = df['Mileage'].fillna(median_mileage)
```

```python
df.isnull().sum()
```

```python
voitures_elec = df['Fuel'] == 'Electric'
if voitures_elec.any():

    elec = df.loc[voitures_elec & (df['Horsepower_kw'] > 0), 'Horsepower_kw']
    if len(elec) > 0:
        moy_elec = elec.mean()
        df.loc[voitures_elec & (df['Horsepower_kw'] == 0), 'Horsepower_kw'] =  moy_elec
```

```python
df['Price (Lakhs)'] < 1
```

```python
df.describe()
```

```python
df['Price (Lakhs)'].describe()
```

```python
df = df.reset_index(drop=True)
```

```python
prix_outliers = df['Price (Lakhs)'] < 1 
if prix_outliers.any():
    print(f"\n Prix suspects (< 1 lakh) trouvés aux lignes : {df[prix_outliers].index.tolist()}")
    print(df[prix_outliers][['Company', 'Model', 'Year', 'Price (Lakhs)']])

    for idx in df[prix_outliers].index:
        m = df[(df['Company'] == df.loc[idx, 'Company']) & 
                          (df['Model'] == df.loc[idx, 'Model'])]['Price (Lakhs)'].median()
        if not pd.isna(m):
            df.loc[idx, 'Price (Lakhs)'] = m
           
```

```python
df.to_csv('CarPriceDataset_Ne.csv', index=False)
```

```python
df['Year'] = pd.to_numeric(df['Year'], errors='coerce')
df['Engine'] = pd.to_numeric(df['Engine'], errors='coerce')
df['Km_driven'] = pd.to_numeric(df['Km_driven'], errors='coerce')
df['Price (Lakhs)'] = pd.to_numeric(df['Price (Lakhs)'], errors='coerce')
df['Horsepower_kw'] = pd.to_numeric(df['Horsepower_kw'], errors='coerce')
```

```python
df = df.dropna(subset=['Price (Lakhs)', 'Year', 'Engine'])
```
